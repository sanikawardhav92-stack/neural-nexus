```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Load cleaned dataset
df = pd.read_csv('/content/cleaned_flood_risk_dataset.csv')

print("Dataset Shape:", df.shape)
df.head()
```

```python
# Target ko binary classification me convert karna
# 0 = No Flood
# 1 = Flood

threshold = 0.5

df['FloodOccurred'] = (
    df['FloodProbability'] >= threshold
).astype(int)

print(df['FloodOccurred'].value_counts())
```

```python
X = df.drop(columns=['FloodProbability', 'FloodOccurred'])

y = df['FloodOccurred']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=500,
    random_state=42,
    class_weight='balanced'
)

rf_model.fit(X_train, y_train)

print("Random Forest Model Training Completed!")
```

```python
y_pred = rf_model.predict(X_test)

print("Predictions:")
print(y_pred[:20])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
f1 = f1_score(y_test, y_pred)

print("F1 Score:", f1)
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['No Flood', 'Flood'],
    yticklabels=['No Flood', 'Flood']
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Random Forest Confusion Matrix")

plt.show()
```
